### Problem 2
#### Write a function that computes (a - b) * c.

```python
def f(a, b, c):
    return (a - b) * c
a, b, c = map(int, input().split())
print(f(a, b, c))
```

### Problem 3
#### Write a function that returns a + b + a * b.

```python
def f(a, b):
    return a + b + a * b
a, b = map(int, input().split())
print(f(a, b))
```

### Problem 4
#### Write a function that returns the count of positive integers k such that 1 <= k < n and k > (n - 1) / 2. (Example: n=9 -> 4).

```python
def f(n):
    return n - 1 - (n - 1) // 2
n = int(input())
print(max(0, f(n)))
```

### Problem 6
#### Write a function that computes the area of a triangle with side lengths a, b, c (use Heron's formula). Raise ValueError if the sides cannot form a triangle.

```python
def f(a, b, c):
    p = (a + b + c) / 2
    area = (p * (p - a) * (p - b) * (p - c)) ** 0.5
    return area
a, b, c = map(int, input().split())
if (a + b > c) and (a + c > b) and (b + c > a):
    print(f(a, b, c))
else:
    print("ValueError")
```

### Problem 7
#### Write a function that returns True if a, b, c can be the side lengths of a triangle; otherwise return False.

```python
def f(a, b, c):
    return (a + b > c and b + c > a and c + a > b)
a, b, c = map(int, input().split())
print(f(a, b, c))
```

### Problem 8
#### Write a function that returns the character that is n positions after character x in the alphabet.
#### Preserve case and wrap around within A..Z or a..z.

```python
def f(n, x):
    return chr(ord(x) + n)
x, n = input().split()
n = int(n)
k = f(n, x)
if k < 'a' or k > 'z':
    print("ValueError")
else:
    print(f(n, x))
```

### Problem 9
#### Write a function that converts an uppercase letter to lowercase. If the input is not a single character, raise ValueError.

```python
def f(x):
    return x.lower()
x = input()
if len(x) != 1 or x[0] < 'A' or x[0] > 'Z':
    print("ValueError")
else:
    print(f(x[0]))
```

### Problem 10
#### Write a function that, given a lowercase letter, returns a string containing that letter repeated 10 times.

```python
def f(x):
    return x * 10
x = input()
if len(x) != 1 or x[0] < '0' or x[0] > '9':
    print("ValueError")
else:
    print(f(x))
```

### Problem 12
#### Write a function that returns a right isosceles triangle of height n using '*'.
#### Line i (1-indexed) has i '*', joined with newlines.
#### Example with n = 4
#### *
#### **
#### ***
#### ****

```python
def f(n):
    res = []
    for i in range(1, n + 1):
        res.append("*" * i)
    return res
n = int(input())
s = f(n)
for i in s:
    print(i)
```
